# check non printable/escape chars

```python
import collections
import re

# Combine all text to analyze as one block
all_text = "".join([str(t) for t in reflection_reports])

# Find all whitespace/escape characters
escapes = re.findall(r'[\n\t\r\xa0]', all_text)
counts = collections.Counter(escapes)

print("--- Escape Character Counts ---")
for char, count in counts.items():
    repr_char = char.encode('unicode_escape').decode()
    print(f"{repr_char}: {count} occurrences")

# Find sequences of special symbols (non-word, non-space characters)
# This will catch things like ==> , --- , ***, etc.
symbols = re.findall(r'[^a-zA-Z0-9\sÀ-ÿ]{2,}', all_text) 
symbol_counts = collections.Counter(symbols)

print("\n--- Symbolic Pattern Counts (Top 15) ---")
for pattern, count in symbol_counts.most_common(15):
    print(f"'{pattern}': {count} occurrences")
```

```python
import pandas as pd
import re
import sys
sys.path.append("/Users/rb666/Projects/MOSAIC/src")
from mosaic.preprocessing.preprocessing import split_sentences

# 1. Define the cleaning logic locally
def clean_text(text):
    if not isinstance(text, str): return ""
    
    # Remove researcher patterns
    patterns = [
        r"==>", 
        r"Note:.*?\n", 
        r"Casus:.*?\n", 
        r"\d{1,2}/\d{1,2}/\d{4}"
    ]
    for p in patterns:
        text = re.sub(p, " ", text, flags=re.IGNORECASE)
        
    # Standardise whitespace and remove the escapes you found (\n, \t, \xa0)
    text = text.replace('\xa0', ' ').replace('\t', ' ').replace('\n', ' ')
    return re.sub(r'\s+', ' ', text).strip()

# 2. Apply cleaning to the raw column
cleaned_reports = [clean_text(report) for report in df_subset['reflection_answer']]

# 3. Perform the split on cleaned data
sentences, doc_map = split_sentences(cleaned_reports)




# 4. Map the IDs (using df_subset to match the reports you processed)
mapped_ids = [df_subset['ID'].iloc[i] for i in doc_map]

# 5. Build the final DF
sentence_df = pd.DataFrame({
    'participant_id': mapped_ids,
    'sentence': sentences
})
```

```python
#sanity check (should be the same number of sentences and mapped ids)
print(len(sentences))
print(len(mapped_ids))

sentence_df
```

```python
#do a supplementary split sentence
def split_sentences_with_punctuation(sentence, participant_id):
    # Split on punctuation marks that are not commas
    sentences = re.split(r'[…!?;:...]+', sentence)
    # Remove any empty strings from the list
    sentences = [s.strip() for s in sentences if s.strip()]
    # Return a list of tuples (sentence, participant_id)
    return [(s, participant_id) for s in sentences]

# Apply the supplementary split to each sentence in sentence_df
new_rows = []
for _, row in sentence_df.iterrows():
    new_rows.extend(split_sentences_with_punctuation(row['sentence'], row['participant_id']))
sentence_df = pd.DataFrame(new_rows, columns=['sentence', 'participant_id'])
#print the new dataframe with split sentences
print("\nDataframe with split sentences:")
len(sentence_df)
```

```python
#check if there are any NaN values in the sentence column
nan_sentences = sentence_df[sentence_df['sentence'].isna()]
print("Sentences with NaN values:")
print(nan_sentences)
```

```python
# Create a boolean mask for rows where both ID and Sentence are identical
# keep=False marks all occurrences as duplicates so you can see them all
duplicates = sentence_df[sentence_df.duplicated(subset=['participant_id', 'sentence'], keep=False)]

# Sort by ID to see the matching pairs together
duplicates_sorted = duplicates.sort_values(by=['participant_id', 'sentence'])

print(f"Found {len(duplicates_sorted)} duplicate rows.")

split_df_cleaned = sentence_df.drop_duplicates(subset=['participant_id', 'sentence'], keep='first')
print(f"After removing duplicates, {len(split_df_cleaned)} unique participant_id-sentence pairs remain.")
```

```python
#now remove all sentences that are shorter than 3 words
split_df_cleaned['word_count'] = split_df_cleaned['sentence'].apply(lambda x: len(str(x).split()))
#print the sentences that are shorter than 3 words
# short_sentences = split_df_cleaned[split_df_cleaned['word_count'] < 3]

final_df = split_df_cleaned[split_df_cleaned['word_count'] >= 3].drop(columns=['word_count'])
print(f"After removing sentences shorter than 3 words, {len(final_df)} sentences remain.")
```
